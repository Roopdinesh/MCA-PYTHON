```python
def is_valid_password (pwd):
    if len(pwd)<8:
        return False
    if not pwd.isalnum():
        return False
    return True

print(is_valid_password("hello"))
print(is_valid_password("helloworld"))
print(is_valid_password("helloworld10"))
print(is_valid_password("200321$123"))
```

```python
# default args

def cleanText(data,to_remove = "@$#%"):
    newdata = data                          # make a copy of data
    for char in to_remove:                  # loop of each char to remove
        newdata = newdata.replace(char,'')  # use replace to remove char
    return newdata                          # return newdata

cText = cleanText("He#ll#o@")
print(cText)
cText2 = cleanText("H@!()#!)@i",to_remove='@!#()') # Updated chars
print(cText2)
```

```python
def multiplier(*numbers):
    result=1
    for value in numbers:
        result *=value
    return value
    
print(multiplier(1))
print(multiplier(2,3,4))
print(multiplier(2,4,6,7,3,9,10,56))
```

```python
def operate(*args,operator='*'):
    if operator=='*':
        result=1
        for num in args:
            result*num
        return result
    elif operator == '+':
        result=0
        for num in args:
            result+=num
        return result
    else:
        return "wrong operator give , please provide + or *"
    
    print(operator(1,2,3,4,))
    print(operator(1,2,3,4,5,6,operate='+'))
    print(operator(1,2,3,4,operate='-'))
```

```python
def squared (data):
    for i in data:
        yield i**2
        
        
for num in squared([2,3,4,3,5,7]):
    print("square",num)
    
print(list(squared([22,33,55])))
```
