```python
# Task 1: Create a Rectangle Class
class Rectangle:
    def __init__(self):
        self.length = 0
        self.width = 0

    def set_dimensions(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)


# Example usage
rect = Rectangle()
rect.set_dimensions(5, 3)

print("Area:", rect.area())
print("Perimeter:", rect.perimeter())

```

```python
# Task 2: Student Grade Manager
class Student:
    def __init__(self, name, mark1, mark2, mark3):
        self.name = name
        self.mark1 = mark1
        self.mark2 = mark2
        self.mark3 = mark3

    def total_marks(self):
        return self.mark1 + self.mark2 + self.mark3

    def average_marks(self):
        return self.total_marks() / 3


# Example usage
student1 = Student("Alice", 85, 90, 80)

print("Student Name:", student1.name)
print("Total Marks:", student1.total_marks())
print("Average Marks:", student1.average_marks())

```

```python
# Task 3: Temperature Converter
class Temperature:
    def __init__(self):
        self.celsius = 0

    def set_celsius(self, celsius):
        self.celsius = celsius

    def to_fahrenheit(self):
        return (self.celsius * 9/5) + 32

    def to_kelvin(self):
        return self.celsius + 273.15


# Example usage
temp = Temperature()
temp.set_celsius(25)

print("Temperature in Celsius:", temp.celsius)
print("Temperature in Fahrenheit:", temp.to_fahrenheit())
print("Temperature in Kelvin:", temp.to_kelvin())
```

```python
# Task 4: Shopping Cart
class ShoppingCart:
    def __init__(self):
        self.items = []  # List to store items as dictionaries

    def add_item(self, name, price):
        self.items.append({'name': name, 'price': price})
        print(f"Added '{name}' for ₹{price}")

    def remove_item(self, name):
        for item in self.items:
            if item['name'] == name:
                self.items.remove(item)
                print(f"Removed '{name}' from the cart")
                break
        else:
            print(f"Item '{name}' not found in the cart")

    def total_price(self):
        return sum(item['price'] for item in self.items)

    def show_cart(self):
        if not self.items:
            print("Your cart is empty.")
        else:
            print("\nItems in your cart:")
            for item in self.items:
                print(f"- {item['name']}: ₹{item['price']}")
            print(f"Total Price: ₹{self.total_price()}")


# Example usage
cart = ShoppingCart()
cart.add_item("Apple", 50)
cart.add_item("Banana", 20)
cart.add_item("Milk", 40)

cart.show_cart()

cart.remove_item("Banana")
cart.show_cart()
```

```python
# Task 5: Employee Management
class Employee:
    def __init__(self):
        self.name = ""
        self.emp_id = 0
        self.salary = 0.0

    def set_details(self, name, emp_id, salary):
        self.name = name
        self.emp_id = emp_id
        self.salary = salary

    def give_raise(self, percentage):
        self.salary *= (1 + percentage / 100)

    def display_info(self):
        print(f"Employee Name: {self.name}")
        print(f"Employee ID: {self.emp_id}")
        print(f"Current Salary: ₹{self.salary:.2f}")


# Example usage
emp = Employee()
emp.set_details("John Doe", 101, 50000)

print("Before Raise:")
emp.display_info()

emp.give_raise(10)  # 10% raise

print("\nAfter 10% Raise:")
emp.display_info()
```
