```python
#1. From Dictionaries

import pandas as pd

# Create DataFrame from dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age': [20, 21, 19, 22],
    'Grade': [85, 78, 92, 88]
}

df = pd.DataFrame(data)
print(df)
#       Name  Age  Grade
# 0    Alice   20     85
# 1      Bob   21     78
# 2  Charlie   19     92
# 3    Diana   22     88

# Check data types
print(df.dtypes)
# Name     object
# Age       int64
# Grade     int64
```
