```python
import matplotlib.pyplot as plt

subjects = ['Python', 'Java', 'WebDev', 'Database']
scores = [85, 78, 92, 88]

plt.bar(subjects, scores, color='skyblue', edgecolor='black')
plt.xlabel('Subjects')
plt.ylabel('Scores')
plt.title('My Grades')
plt.ylim(0, 100)
plt.show()
```

```python
#Horizontal Bars:

import matplotlib.pyplot as plt

subjects = ['Python', 'Java', 'WebDev', 'Database']
scores = [85, 78, 92, 88]

plt.barh(subjects, scores, color='lightcoral')
plt.xlabel('Scores')
plt.ylabel('Subjects')
plt.title('My Grades (Horizontal)')
plt.xlim(0, 100)
plt.show()
```
