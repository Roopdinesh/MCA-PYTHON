```python
import matplotlib.pyplot as plt

subjects = ['Python', 'Java', 'WebDev', 'Database']
scores = [85, 78, 92, 88]

plt.bar(subjects, scores, color='skyblue', edgecolor='black')
plt.xlabel('Subjects')
plt.ylabel('Scores')
plt.title('My Grades')
plt.ylim(0, 100)
plt.show()
```

```python
temperature=[35.5,32.5,27.5,28.5,26.9,20.4,24.2]
days=['monday','tuesday','wednesday', 'thursday', 'friday','saturday','sunday']
plt.bar(days,height=temperature)
```

```python
 #Horizontal Bars:

import matplotlib.pyplot as plt

subjects = ['Python', 'Java', 'WebDev', 'Database']
scores = [85, 78, 92, 88]

plt.barh(subjects, scores, color='lightcoral')
plt.xlabel('Scores')
plt.ylabel('Subjects')
plt.title('My Grades (Horizontal)')
plt.xlim(0, 100)
plt.show()
```
