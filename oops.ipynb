```python
class pokemon:
    def init(self,name,type='normal',hp=10,atk=10,dnf=10):
        self.name=name
        self.type=type
        self.hp=hp
        self.atk=atk
        self.dnf=dnf
    def stats(self):
        return(f'Name:{self.name}\nType:{self.type}\nHP:{self.hp}\nATK:{self.atk}\nDNF:{self.dnf}')
```

```python
pikachu = pokemon("Pikachu", "Electric",35, 55, 40)
snorlax = pokemon("Snorlax", "Normal", 160, 110, 65)
charizard = pokemon("Charizard", "Fire/Flying", 78, 84, 78)
bulbasaur = pokemon("Bulbasaur", "Grass/poison", 45, 49, 49)
rattata = pokemon("Rattata")

pokemons = [pikachu, snorlax, charizard, bulbasaur, rattata]
print("\nStats")
for poke in pokemons:
    print(poke.stats())
   
    print('-------------')
```

```python
class player:
    def __init__(self,name,score):        #__init__ is a pre defined method
        self.name=name
        self.score=score
    def __gt__(self,other):
        return self.score>other.score
        
```

```python
p1=player('sam',20)
p2=player('ram',30)
if p1>p2:
    print('f"{p1.name} is better')
else:
    print(f"{p2.name} is better")
```

```python
class player:
    def __init__(self,name,score):        #__init__ is a pre defined method
        self.name=name
        self.score=score
    def __gt__(self,other):
        if isinstance(other.player):
            return self.score>other.score
        
```
